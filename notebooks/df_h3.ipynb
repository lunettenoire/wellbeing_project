```python
import pandas as pd

# 1. Charger le dataset à partir de l'URL
url = "https://raw.githubusercontent.com/lunettenoire/wellbeing_project/refs/heads/main/data/cleaned/df_h1.csv"
df = pd.read_csv(url)

# 2. Ne garder que les colonnes souhaitées
colonnes_a_garder = ['country_name', 'log_gdp_per_capita', 'year', 'positive_affect', 'negative_affect']
df_filtre = df[colonnes_a_garder]

# 3. Afficher le nombre total de valeurs manquantes par colonne
print("Nombre de valeurs manquantes par colonne :")
print(df_filtre.isnull().sum())
print("-" * 50)

# 4. Filtrer et afficher toutes les lignes qui contiennent au moins une valeur manquante
lignes_manquantes = df_filtre[df_filtre.isnull().any(axis=1)]

print(f"Nombre de lignes avec au moins une valeur manquante : {len(lignes_manquantes)}")
lignes_manquantes
```

```python
nb_pays_uniques = df['country_name'].nunique()

print(f"Il y a {nb_pays_uniques} pays uniques dans le jeu de données.")
```

```python
# Liste des colonnes concernées
colonnes = ["positive_affect", "negative_affect"]

# Calcul de la moyenne entre la ligne 1078 et la ligne 1080
moyenne_prec_suiv = df.loc[[1078, 1080], colonnes].mean()

# Remplacement des valeurs manquantes à l'index 1079
df.loc[1079, colonnes] = df.loc[1079, colonnes].fillna(moyenne_prec_suiv)
```

```python
# Suppression des lignes spécifiées
df.drop([488, 489, 490], axis=0, inplace=True)
```

```python
# Enregistrer le DataFrame dans un nouveau fichier CSV
df.to_csv("df_h3", index=False)
```
