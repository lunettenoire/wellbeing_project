```python
import pandas as pd
from scipy.stats import shapiro, spearmanr

# 1. Chargement de votre fichier de données
df = pd.read_csv('https://raw.githubusercontent.com/lunettenoire/wellbeing_project/refs/heads/main/data/cleaned/h2_clean_corruption.csv')

# Liste des variables à analyser face au bonheur ('life_ladder')
variables_a_tester = [
    'social_support', 
    'freedom_to_make_life_choices', 
    'generosity', 
    'perceptions_of_corruption'
]

# Inclure également 'life_ladder' pour vérifier sa propre distribution
toutes_les_variables = ['life_ladder'] + variables_a_tester


print("=== ÉTAPE 1 : VÉRIFICATION DE LA NORMALITÉ (SHAPIRO-WILK) ===")
print("Objectif : Savoir si les données suivent une loi normale (courbe en cloche).\n")

# Dictionnaire pour stocker si la variable est normale ou non
est_normale = {}

for var in toutes_les_variables:
    # On retire les valeurs manquantes pour le test de Shapiro
    donnees_propres = df[var].dropna()
    
    # Calcul du test de Shapiro-Wilk
    statistique, p_val_shapiro = shapiro(donnees_propres)
    
    print(f"• Variable : {var}")
    print(f"  - P-value de Shapiro : {p_val_shapiro:.3e}")
    
    # Interprétation du test de normalité
    if p_val_shapiro < 0.05:
        print("  -> Conclusion : La p-value est < 0.05. La distribution n'est PAS normale.")
        est_normale[var] = False
    else:
        print("  -> Conclusion : La p-value est > 0.05. La distribution est normale.")
        est_normale[var] = True
    print("-" * 60)

print("\nJUSTIFICATION DU CHOIX DU TEST :")
print("Comme les variables ne suivent pas une loi normale, le test de Pearson serait faussé.")
print("Nous utilisons donc obligatoirement le test de Spearman (basé sur les rangs).\n")


print("=== ÉTAPE 2 : CALCUL DES CORRÉLATIONS DE SPEARMAN ===\n")

for var in variables_a_tester:
    # Nettoyage des lignes vides pour le couple de variables
    donnees_couple = df[['life_ladder', var]].dropna()
    
    # Calcul du coefficient de Spearman
    rho, p_value_spearman = spearmanr(donnees_couple['life_ladder'], donnees_couple[var])
    
    print(f"• Lien entre 'life_ladder' et '{var}' :")
    print(f"  - Coefficient de corrélation (rho) : {rho:.3f}")
    print(f"  - P-value du test : {p_value_spearman:.3e}")
    
    # Interprétation textuelle du résultat
    if p_value_spearman < 0.05:
        if rho > 0.6:
            print(f"  -> Interprétation : LIEN POSITIF TRÈS FORT. Plus '{var}' augmente, plus le bonheur augmente.")
        elif rho > 0.3:
            print(f"  -> Interprétation : LIEN POSITIF MODÉRÉ. Plus '{var}' augmente, plus le bonheur tend à augmenter.")
        elif rho > 0:
            print(f"  -> Interprétation : LIEN POSITIF FAIBLE. Légère tendance à la hausse du bonheur.")
        elif rho < -0.3:
            print(f"  -> Interprétation : LIEN NÉGATIF MODÉRÉ. Plus '{var}' augmente, plus le bonheur diminue.")
        else:
            print(f"  -> Interprétation : LIEN NÉGATIF FAIBLE. Légère tendance à la baisse du bonheur.")
    else:
        print(f"  -> Interprétation : Aucun lien statistiquement significatif.")
    print("-" * 60)
```
